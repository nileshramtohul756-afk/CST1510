### D6. A default value

1. Define `status_of(percent, warning_at=90)` that returns:
   - `"OVER LIMIT"` if `percent` is 100 or more
   - `"WARNING"` if `percent` is `warning_at` or more
   - `"OK"` otherwise
2. Print `status_of(95)`, then `status_of(85)`, then `status_of(85, 80)`.

**Expected output:** `WARNING`, `OK`, `WARNING`

````python
# D6
# defines the function names status_of that takes two parameters:
# percent and warning_at which has a default value of 90
def status_of(percent, warning_at = 90):
    if percent >= 100:
        return("OVER LIMIT")
    elif percent >= warning_at:
        return("WARNING")
    else:
        return("OK")

print(status_of(95))
print(status_of(85))
print(status_of(85,80))
````

````python
# D8
# defines a function named total that takes a variable number of arguments
def total(*numbers):
    return sum(numbers) # returns the sum of arguments using built in sum function

print(total(1,2,3))
print(total(10,20,30,40))
print(total())

# defines a function named configure that takes a variable number of keyword arguments
def configure(**settings):
    for key, value in settings.items(): # loops over the items in  the settings dictionary
        print(f"{key} = {value}")


configure(host="10.0.0.1", port=8080)
````

### D12. Returning more than one value

1. Define `check_record(value, limit)` that works out `difference = value - limit` and
   `percent = (value / limit) * 100`, and returns **both**: `return difference, percent`
2. Call it with `check_record(87.0, 100.0)` and store the results in two variables.
3. Print them.

**Expected output:**
```
Difference: -13.0
Percent: 87.0
```

````python
# D12
# defines a function called check_record that takes two parameters, value and limit
def check_record(value,limit):
    difference = value - limit  # calculates difference between value and limit
    percent = (value/limit) * 100 # calculates value as a percentage of the limit
    return difference, percent

# call the check_record function with 87.0 and 100.0 , storing the results in two variables
difference, percent = check_record(87.0, 100.0) 

print(f"Difference : {difference}")
print(f"Percent : {percent}")
````

### C1. Wire two functions together

Using `percent_of` from D3 and `status_of` from D6 **together**, print a status for each of these
three pairs. Do not write the percentage or status logic a second time.

| value | limit | Expected output |
|---|---|---|
| 87 | 100 | `OK` |
| 120 | 100 | `OVER LIMIT` |
| 45 | 100 | `OK` |

````python
# C1
def percent_of(part,total):  # defines a function named percent_of that takes part and
                             # total as parameters
    return (part/total) * 100 # returns part as a percentage of total

def status_of(percent, warning_at = 90): 
    if percent >= 100:
        return "OVER LIMIT"
    elif percent >= warning_at:
        return "WARNING"
    else:
        return "OK"

print(status_of(percent_of(87,100)))
print(status_of(percent_of(120,100)))
print(status_of(percent_of(45,100)))

````

### C2. How many are over?

Write `count_over(*values, limit=100)` with a one-line docstring. It returns how many of the values
are over the limit.

| Call | Expected output |
|---|---|
| `count_over(87, 120, 45, 101)` | `2` |
| `count_over(87, 120, 45, 101, limit=50)` | `3` |
| `count_over()` | `0` |

````python
# C2
def count_over(*values, limit = 100):
    """Return the number of values that are greater than the limit"""
    return len([v for v in values if v > limit]) # len() is used to count the numbers greater than the limit

print(count_over(87, 120, 45, 101))
print(count_over(87, 120, 45, 101, limit=50))
print(count_over())
````
