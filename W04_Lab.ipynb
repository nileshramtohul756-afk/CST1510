### D9. Loop over a dict

Using `record` from D7, write three loops:

a) print only the **keys** (`for key in record:`)

b) print only the **values** (`.values()`)

````python
# D9
record = {
    "label": "R-001",
    "value": 87.0,
    "limit": 100.0
}

for key in record: # loops through the dictionary and prints the keys only
    print(key)

for value in record.values():  # loops through the dictionary and prints the values only
    print(value)

for key,value in record.items(): 
    print(f"{key} : {value}")

    
````

### D14. Catch a `ValueError`

1. Create `text = "abc"`.
2. Inside `try:`, convert it with `float(text)` and print the number.
3. In `except ValueError:`, print `Not a number: abc` instead.
4. Now change `text` to `"42"` and run again.

**Expected output:** `Not a number: abc`, then (after the change) `42.0`

````python
# D14
text = "abc"
try:
    print(float(text)) # attempt to convert string to a float 
except ValueError: 
    print("Not a number: abc")
 
````

````python
text = "42"
try:
    print(float(text))
except ValueError: # catches the ValueError and prevents crash if string cannot be converted to a float
    print("Not a number: abc")
````

### D15. Catch a `FileNotFoundError`

1. Inside `try:`, open `missing.txt` with `with` and print its contents.
2. In `except FileNotFoundError:`, print `missing.txt not found`.

**Expected output:** `missing.txt not found`

````python
# D15
try: 
    with open("missing.txt") as f: # attempts to read a file that does not exist
        content = f.read()
        print(content) 
except FileNotFoundError: # catches the named error to prevent crashing when file is not found
    print("missing.txt not found")               
````

### C1. Load records safely from a file

Write `load_records(path)` with a one-line docstring. It reads `label,value,limit` lines from a file
and returns a **list of dicts**. If the file does not exist, it prints a clear message and returns an
empty list instead of crashing.

Then load `records_sample.txt` and print each record's label and status, as in D16.

**Test it:**

| Call | Expected output |
|---|---|
| `load_records("records_sample.txt")` | the same three lines as D16 |
| `load_records("nope.txt")` | a clear message, then `[]` |

````python
# C1
def load_records(path):
    """Reads label,value,limit lines from a file and returns a list of dicts"""
    records_lists = [] # creates an empty list to hold the dictionaries 

    try:
        with open(path, "r") as f: # opens the file in read mode
            for line in f:
                label, value, limit = line.strip().split(",")
                records_lists.append({ # creates a dictionary for the line and appends it to the list
                    "label" : label,
                    "value" : float(value),
                    "limit" : float(limit)
                })
        return records_lists
    except FileNotFoundError: # catches the named error to prevent crashing when file is not found
        print(f"{path} not found")  
        return []


loaded_records = load_records("records_sample.txt") 
for record in loaded_records:
    if record["value"] > record["limit"]:
        status = "OVER LIMIT"
    else:
        status = "OK"   

    print(f"{record["label"]} : {status}")
````
