## Topic 2 · Methods
*Walkthrough: `02 - Methods`*

### D5. Add a method

1. Add a method `summary(self)` to the class below. It **returns** a string like `R-001: 87.0/100.0`,
   built from `self.label`, `self.value` and `self.limit`.
2. Create a record and print the result of calling `.summary()` on it.

**Expected output:** `R-001: 87.0/100.0`

````python
# D5
class Record:
    def __init__(self, label, value, limit):
        self.label = label
        self.value = value
        self.limit = limit

    def summary(self):# defines a new method inside the class that uses self to access other attributes
        return f"{self.label} : {self.value}/{self.limit}"

r = Record("R-001", 87.0, 100.0)
print(r.summary()) # calls the summary method
````

### D6. `self` is whichever object made the call

1. Copy your class from D5 and add a method `status(self)` that returns `"OVER LIMIT"` (100% or more),
   `"WARNING"` (90% or more) or `"OK"`, worked out from `self.value` and `self.limit`.
2. Create three records and print each one's `.status()`:

| Record | Expected output |
|---|---|
| `Record("R-001", 87.0, 100.0)` | `OK` |
| `Record("R-002", 120.0, 100.0)` | `OVER LIMIT` |
| `Record("R-003", 95.0, 100.0)` | `WARNING` |

````python
# D6
class Record:
    def __init__(self, label, value, limit):
        self.label = label
        self.value = value
        self.limit = limit

    def status(self):
        """Return value as a percentage of limit"""
        p = (self.value / self.limit) * 100
        if p >= 100:
            return "OVER LIMIT"
        elif p >= 90:
            return "WARNING"
        else:
            return "OK"

# creates three record objects
r1 = Record("R-001", 87.0, 100.0)
r2 = Record("R-002", 120.0, 100.0)
r3 = Record("R-003", 95.0, 100.0)

# calls the .status() method and prints the status for each object
print(f" r1 = {r1.status()}")
print(f" r2 = {r2.status()}")
print(f" r3 = {r3.status()}")
````

### D7. One method calls another

1. Copy your class from D6 and add a method `percent(self)` that returns `(self.value / self.limit) * 100`.
2. Change `status(self)` so it uses `self.percent()` instead of working out the percentage itself.
3. Create `Record("R-001", 87.0, 100.0)` and print its `.percent()` and `.status()`.

**Expected output:** `87.0`, then `OK`

````python
# D7
class Record:
    def __init__(self, label, value, limit):
        self.label = label
        self.value = value
        self.limit = limit


    def percent(self):
        """Returns the value as a percentage of the limit"""
        return (self.value)/(self.limit) * 100 # calculates and returns percent

    def status(self):
        """Returns OVER LIMIT, WARNING or OK depending on value of percent"""
        p = self.percent() # calls the object's percent() method using "self"
        if p >= 100:
            return "OVER LIMIT"
        elif p >= 90:
            return "WARNING"
        else:
            return "OK"


r = Record("R-001", 87.0, 100.0)
print(f"{r.percent()}, {r.status()}")
````

### D8. A method can change its own object

1. Define a class `Checker`. In `__init__(self)`, set `self.over_limit_count = 0`.
2. Add a method `check(self, value, limit)` that adds 1 to `self.over_limit_count` when `value` is
   greater than `limit`.
3. Create one `Checker`, call `.check()` three times with `(87, 100)`, `(120, 100)` and `(130, 100)`,
   then print `over_limit_count`.

**Expected output:** `2`

````python
# D8
class Checker:
    def __init__(self):
        self.over_limit_count = 0

    def check(self, value, limit):
        """When value > limit, over limit count increases by 1"""
        if value > limit:   
            self.over_limit_count += 1

checker_a = Checker()   

# calls the check method three times with different values
checker_a.check(87,100)
checker_a.check(120,100)
checker_a.check(130,100)
print(checker_a.over_limit_count) # prints the final count

       
````

### D9. A list of objects

1. Using your `Record` class from D7, create a list of three records:
   `R-001` (87.0), `R-002` (120.0) and `R-003` (45.0), each with a limit of 100.0.
2. Loop over the list and print each record's label and status.

**Expected output:**
```
R-001 OK
R-002 OVER LIMIT
R-003 OK
```

````python
# D9
class Record:
    def __init__(self, label, value, limit):
        self.label = label
        self.value = value
        self.limit = limit


    def percent(self):
        """Returns the value as a percentage of the limit"""
        return (self.value)/(self.limit) * 100

    def status(self):
        """Returns OVER LIMIT, WARNING or OK depending on value of percent"""
        p = self.percent()
        if p >= 100:
            return "OVER LIMIT"
        elif p >= 90:
            return "WARNING"
        else:
            return "OK"


records = [ # creates a list containing multiple Record objects
     Record("R-001", 87.0, 100.0),
     Record("R-002", 120.0, 100.0),
     Record("R-003", 45.0, 100.0)
]            

for r in records: # iterates through the list and calls the method on each object
     print(r.label, r.status())
````

### D11. Load objects from a file

`records_sample.txt` (in this `LAB` folder) has one record per line, like `R-001,87,100`.

1. Write `load_records(path)` that reads the file and returns a list of `Record` objects
   (convert `value` and `limit` with `float()`).
2. Loop over the list and print each record's label and status.

**Expected output:** the same three lines as D9.

````python
# D11
def load_records(path):
    """Read label,value,limit lines from a file and return a list of Records."""
    records = []
    with open(path) as f:
        for line in f: 
            label,value,limit = line.strip().split(",") # takes the data from line
            records.append(
                Record((label), 
                float(value), 
                float(limit)))
    return records        

def status(self):
        """Returns OVER LIMIT, WARNING or OK depending on value of percent"""
        p = self.percent()
        if p >= 100:
            return "OVER LIMIT"
        elif p >= 90:
            return "WARNING"
        else:
            return "OK"
        
records_a = load_records("records_sample.txt")

for record in records_a:
  print(f"{record.label} {record.status()}")
````

### C1. Count the records over the limit

Add a method `is_over(self)` to `Record` that returns `True` or `False`. Load `records_sample.txt` with
`load_records` from D11, use `is_over()` to count how many records are over their limit, and print a
one-line summary.

**Expected output:** `1 of 3 records are OVER LIMIT`

````python
# C1
class Record:
    def __init__(self, label, value, limit):
        self.label = label
        self.value = value
        self.limit = limit

    def percent(self):
        """Returns the value as a percentage of the limit"""
        return (self.value)/(self.limit) * 100

    def status(self):
        if self.is_over():
            return "OVER LIMIT"
        else:
            return "OK"

    def is_over(self):
         """Returns True if percent is over 100, otherwise False"""
         return self.percent() >= 100


def load_records(path):
    """Read label,value,limit lines from a file and return a list of Records."""
    records = []
    with open(path) as f:
        for line in f:
            label,value,limit = line.strip().split(",")
            records.append(Record(label, float(value), float(limit)))
    return records        


records_a = load_records("records_sample.txt")

over_limit_count = 0 
total_records = len(records_a)

# is_over is used to count how many records are over the limit
for record in records_a:
        if record.is_over():
          over_limit_count += 1



print(f"{over_limit_count} of {total_records} records are OVER LIMIT")        






````

### C2. An account that locks

Write a class `Account` with a `username` and a count of failed logins that starts at 0.
It has two methods:

- `fail(self)`: adds 1 to the failed logins
- `locked(self)`: returns `True` if there have been 3 or more failed logins

**Test it:**

| Code | Expected output |
|---|---|
| an account for `"anna"` with `.fail()` called 3 times, then `print(anna.locked())` | `True` |
| an account for `"ben"` with `.fail()` called once, then `print(ben.locked())` | `False` |

````python
# C2
class Account:
    def __init__(self, username):
        self.username = username
        self.failed_logins = 0

    def fail(self):     # Adds 1 to the failed logins
        self.failed_logins += 1

    def locked(self): # Returns True if failed logins is 3 or more, otherwise False
        return self.failed_logins >= 3
            
            
````

````python
# test
anna = Account("anna")
anna.fail()
anna.fail()
anna.fail()
print(anna.locked())
````

````python
ben = Account("ben")
ben.fail()
print(ben.locked())
````
